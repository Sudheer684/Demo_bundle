```python
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
table = dbutils.widgets.get("table")
full_table_name = f"{catalog}.{schema}.{table}"
```

```python
# Sample Python code
from pyspark.sql import SparkSession

# Create a small DataFrame
spark = SparkSession.builder.getOrCreate()
df = spark.createDataFrame(
    [(1, "apple", 10.0), (2, "banana", 5.5), (3, "cherry", 8.25)],
    schema="id INT, fruit STRING, price DOUBLE",
)

# Basic transformations
df_filtered = df.filter(df.price > 6.0) \
    .withColumn("price_with_tax", df.price * 1.08)

df_filtered.show()

# Simple pandas example
data = {"x": [1, 2, 3, 4], "y": [10, 20, 25, 30]}
print(sum(data["x"]), sum(data["y"]) / len(data["y"]))
```

```python
# Save the DataFrame as a Delta table in a Unity Catalog schema
# catalog / schema / table are set with the notebook widgets at the top of the notebook
# (also usable as job parameters, e.g. when run from a job)


df.write \
    .mode("overwrite") \
    .saveAsTable(full_table_name)

print(f"Table written to {full_table_name}")
spark.table(full_table_name).show()
```
