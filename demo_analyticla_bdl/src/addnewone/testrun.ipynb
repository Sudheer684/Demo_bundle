```python
# 1.conftest.py
import pytest
from pyspark.sql import SparkSession
@pytest.fixture(scope="session")
def spark():
    return SparkSession.builder.getOrCreate()

# 2.transformations.py
from pyspark.sql.functions import sum
def total_amount_per_customer(df):
    return df.groupBy("customer_id") \
             .agg(sum("amount").alias("total_amount"))


# 3.test_transformations.py
from transformations import total_amount_per_customer
def test_total_amount_per_customer(spark):
    data = [
        (1, 100),
        (1, 400),
        (2, 300)
    ]
    columns = ["customer_id", "amount"]

    df = spark.createDataFrame(data, columns)
    result = total_amount_per_customer(df)

    assert {r["customer_id"]: r["total_amount"] for r in result.collect()} == {
        1: 300,
        2: 300
    }
```
