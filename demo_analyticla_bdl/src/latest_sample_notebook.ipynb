```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG `{catalog}`")
spark.sql(f"USE SCHEMA `{schema}`")
```

```python
# Create a simple Delta table in Unity Catalog.
# Update the catalog, schema, and table name to fit your workspace.
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{schema}.student (
    id BIGINT,
    name STRING,
    created_at TIMESTAMP
)
""")
```
